# Download VIDA building footprints for Uruguay and filter by AOI

```python
from pathlib import Path
import urllib.request
import geopandas as gpd

URL = "https://data.source.coop/vida/google-microsoft-open-buildings/geoparquet/by_country/country_iso=URY/URY.parquet"
PARQUET = Path("data/URY.parquet")
OUT = Path("data/buildings_aoi.gpkg")
```

```python
if not PARQUET.exists():
    urllib.request.urlretrieve(URL, PARQUET)
print(f"{PARQUET.stat().st_size / 1e6:.0f} MB")
```

```python
aoi = gpd.read_file("aoi.gpkg")
buildings = gpd.read_parquet(PARQUET, bbox=tuple(aoi.total_bounds)).to_crs(aoi.crs)
buildings_aoi = buildings[buildings.within(aoi.union_all())]
print(f"{len(buildings)} buildings in AOI bbox, {len(buildings_aoi)} within AOI")
```

```python
buildings_aoi = buildings_aoi[buildings_aoi["area_in_meters"] >= 15]
print(f"{len(buildings_aoi)} buildings of 15 m² or more")
```

```python
buildings_aoi.to_file(OUT, layer="buildings")
buildings_aoi.head(5)
```
